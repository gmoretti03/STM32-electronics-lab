Interpolazione misure SiPM tensione di Breakdown in funzione del guadagno

```python
!pip install iminuit

import numpy as np
import matplotlib.pyplot as plt
from iminuit import Minuit
from iminuit.cost import LeastSquares
```

```python
def f (x, m, q):
    return m*x +q
```

```python
x = [1,2,3,4,5]
y = [1.5949, 3.0783, 4.6853, 6.5465, 8.3216]
err = [0.25168, 0.51165, 0.96256, 1.617, 2.5923]
```

```python
least_squares = LeastSquares (x, y, err, f)
my_minuit = Minuit (least_squares, m = 0, q = 0)
my_minuit.migrad ()
my_minuit.hesse ()
```

```python
print (my_minuit.values[1], my_minuit.errors[1])
```

```python
 t = - my_minuit.values[1] / my_minuit.errors[1]
```

```python
print ('valore di t:',t)
```

```python
Q_squared = my_minuit.fval
print ('value of the fit Q-squared', Q_squared)
N_dof = my_minuit.ndof
print ('value of the number of degrees of freedom', N_dof)
```
